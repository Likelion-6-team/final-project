### 데이터 조회 함수

```python
# 데이터 확인 함수

def get_data_spec(df):
    null_ratio = (df.isnull().mean() * 100).round(2)
    spec = pd.DataFrame({
        '변수명': df.columns,
        '데이터 타입': df.dtypes.values,
        '결측치 개수': df.isnull().sum().values,
        '결측치 비율': [f"{val}%" for val in null_ratio.values]
    })
    return spec
```
