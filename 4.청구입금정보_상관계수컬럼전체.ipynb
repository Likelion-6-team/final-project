```python
import pandas as pd
import numpy as np
from dython.nominal import associations
from sklearn.preprocessing import LabelEncoder
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_parquet('1.회원정보_train.parquet')
df1 = pd.read_parquet('4.청구입금정보_train.parquet')

df
```

```python
# LabelEncoder 하기

object_col = ['ID','대표결제방법코드','대표청구지고객주소구분코드','대표청구서수령지구분코드','청구서수령방법','할인건수_R3M','할인건수_B0M']

le = LabelEncoder()

for col in object_col:
    df1[col] = le.fit_transform(df1[col].astype(str))
```

```python
# LabelEncoder 하기
le = LabelEncoder()
df['Segment'] = le.fit_transform(df['Segment'].astype(str))
```

```python
# 피어슨 상관계수 구하기

segment = df['Segment']

correlations = df1.corrwith(segment, method = 'pearson')
```

```python
significant_corr = correlations[correlations.abs() >= 0.1]
print(significant_corr)
```
