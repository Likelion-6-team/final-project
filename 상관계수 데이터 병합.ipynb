```python
import pandas as pd
import glob
import os

# 1. 병합 대상 경로 설정
csv_dir = "./corr"
csv_files = glob.glob(os.path.join(csv_dir, "*.csv"))

# 2. CSV 파일 읽기 (인코딩 자동 대응)
df_list = []
for file in csv_files:
    try:
        df = pd.read_csv(file, encoding="utf-8")
    except UnicodeDecodeError:
        df = pd.read_csv(file, encoding="cp949")  # 또는 euc-kr
    df_list.append(df)

# 3. 하나로 병합
merged_df = pd.concat(df_list, ignore_index=True)

# 4. 저장 (선택)
merged_df.to_csv("./corr/모든_상관계수_결측치_제거.csv", index=False)
```

```python
df = merged_df
df
```

```python
df.to_parquet("./corr/모든_상관계수_결측치_제거.parquet", index=False)
```
