```python
import pandas as pd

from xgboost import XGBClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report
from sklearn.model_selection import cross_val_score
from sklearn.metrics import make_scorer, f1_score
from sklearn.model_selection import GridSearchCV
```

```python
train = pd.read_csv('./result/상위5개컬럼모음.csv')
# test = pd.read_csv('상위6개컬럼모음.test.csv')
```

```python
X_train=train.drop(['Segment'],axis=1)
y=train['Segment']
# X_test = test
```

```python
y.value_counts().sort_index()
```

```python
X_train.info()
```

```python
# 문자열 클래스일 경우 숫자로 변환
le = LabelEncoder()
y = le.fit_transform(y)
```

```python
# 데이터 분리하기
X_train, X_val, y_train, y_val = train_test_split(X_train, y, test_size=0.2)
print(X_train.shape, X_val.shape, y_val.shape, y_train.shape)
```

```python
XGBClassifier(tree_method="hist", device="cuda")
```

```python
# 학습
model = XGBClassifier(
    objective='multi:softmax',
    num_class=len(le.classes_),
    # use_label_encoder=False,
    eval_metric='mlogloss',
    tree_method="hist", 
    device="cuda"
)

model.fit(X_train, y_train)
y_val_train = model.predict(X_val)
```

```python
# 교차검증
f1_micro = make_scorer(f1_score, average='micro')
scores = cross_val_score(model, X_train, y_train, cv=5, scoring=f1_micro)
```

```python
print("교차검증 평균 F1:", scores.mean())
```

```python
param_grid = {
    'max_depth': [3, 5, 7],
    'learning_rate': [0.01, 0.1, 0.3],
    'n_estimators': [100, 300]
}

grid = GridSearchCV(
    estimator=XGBClassifier(objective='multi:softmax', num_class=5, eval_metric='mlogloss'),
    param_grid=param_grid,
    scoring='f1_macro',
    cv=3
)

grid.fit(X_train, y_train)

print("최적 F1:", grid.best_score_)
print("최적 파라미터:", grid.best_params_)
```

```python
y_val_pred = model.predict(X_val)
print("최종 F1 (micro):", f1_score(y_val, y_val_pred, average='micro'))
print(classification_report(y_val, y_val_pred, target_names=le.classes_))
```

### VIP

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.cross_decomposition import PLSRegression
from sklearn.preprocessing import LabelEncoder, StandardScaler

# 그래프 기본 설정
plt.rcParams['font.family'] = 'Malgun Gothic'
# plt.rcParams['font.family'] = 'AppleGothic'
plt.rcParams['figure.figsize'] = 12, 6
plt.rcParams['font.size'] = 14
plt.rcParams['axes.unicode_minus'] = False
```

```python
# 데이터 불러오기
df = pd.read_csv('./result/상위5개컬럼모음.csv', encoding='utf-8-sig')
X = df.drop(columns=['Segment'])
y = df['Segment']

# 상위 10,000개 행만 사용
df_sampled = df.sample(n=10000, random_state=42)

X = df_sampled.drop(columns=['Segment'])
y = df_sampled['Segment']

# 인코딩 & 정규화
le = LabelEncoder()
y_encoded = le.fit_transform(y)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# PLS 모델 및 VIP 계산
pls = PLSRegression(n_components=2)
pls.fit(X_scaled, y_encoded)

def calculate_vip(pls_model, X):
    t = pls_model.x_scores_
    w = pls_model.x_weights_
    q = pls_model.y_loadings_
    p, h = w.shape
    ssq = np.sum(np.square(t) @ np.square(q.T), axis=0)
    total_ssq = np.sum(ssq)
    vip = np.sqrt(p * (np.sum((np.square(w) * ssq), axis=1)) / total_ssq)
    return vip

vip_scores = calculate_vip(pls, X_scaled)

vip_df = pd.DataFrame({
    'Feature': X.columns,
    'VIP Score': vip_scores
}).sort_values(by='VIP Score', ascending=False)
```

```python
# VIP 점수 정렬
vip_df_sorted = vip_df.sort_values(by="VIP Score", ascending=False)

# 시각화
plt.figure(figsize=(8, 4))
plt.barh(vip_df_sorted['Feature'], vip_df_sorted['VIP Score'])
plt.axvline(1.0, color='r', linestyle='--', label='VIP = 1.0 기준선')
plt.xlabel("VIP Score")
plt.title("PLS Variable Importance (VIP)")
plt.legend()
plt.gca().invert_yaxis()  # 점수 높은 게 위로
plt.tight_layout()
plt.show()
```
